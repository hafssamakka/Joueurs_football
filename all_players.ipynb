## 2 · Univarié

#### Afficher la distribution de la note générale OVR.

- le meilleur graphique pour visualiser la distribution de la note OVR est l'histogramme parce que l'OVR est une variable quantitative continue.

```python
plt.figure(figsize=(12,8))
sns.histplot(df["OVR"], bins=30, kde=True, alpha=0.7) # alpha=0.7 ajuste la transparence des barres de l'histogramme.
plt.xlabel("OVR")
plt.ylabel("Nombre de joueurs")
plt.title("La distribution de OVR des joueurs")
plt.show()
```

- on observe que la distribution de OVR est asymétrique à gauche, ce qui signifie que la majorité des joueurs ont un OVR inférieur à la moyenne, tandis qu'un petit nombre de joueurs ont un OVR très élevé. Cela peut indiquer que la plupart des joueurs sont de niveau moyen ou faible, tandis qu'un petit nombre de joueurs sont très talentueux et ont un OVR élevé.

#### Comparer la distribution de PAC (vitesse) et de PHY (physique) dans deux histogrammes côte à côte.

- le graphique utilisé est un histogramme avec une estimation de la densité de probabilité (KDE) superposée , parce que l'histogramme est utile pour visualiser la distribution des données discrètes, tandis que la courbe KDE fournit une estimation lisse de la densité de probabilité. Cela permet de mieux comprendre la forme générale de la distribution des valeurs de "OVR" et d'identifier les tendances ou les caractéristiques importantes dans les données.


```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4), sharex=True, sharey=True)
colors = ["#aa7b2a", "#62ae6a"]
sns.histplot(df, x="PAC", bins=30, ax=axes[0],kde=True, color=colors[0]).set_title("Vitesse (PAC)")
sns.histplot(df, x="PHY", bins=30, ax=axes[1],kde=True, color=colors[1]).set_title("Physique (PHY)")
plt.show()
```

- On peut voir que la distribution de la vitesse est plus concentrée vers les valeurs élevées, tandis que la distribution du physique est plus étalée avec une concentration plus importante vers les valeurs moyennes. Cela suggère que les joueurs ont tendance à avoir des vitesses plus élevées que des niveaux de physique élevés.

#### Dessiner une densité (kdeplot) pour DRI (dribble).

```python
plt.figure(figsize=(12,8))
sns.kdeplot(df, x="DRI", fill=True, cut=0)
plt.title("Densité de DRI des joueurs")
plt.ylabel("Densité")
plt.xlabel("DRI")
```

- la densité du DRI est concentrée autour de 60-70, ce qui suggère que la plupart des joueurs ont un niveau de dribble moyen à élevé. Il y a également une petite concentration autour de 90, ce qui indique qu'il y a quelques joueurs avec un niveau de dribble exceptionnellement élevé.

#### Quels sont les championnats les plus représentés (League) ?

- Le graphique choisi est:  countplot. il est utilisé pour afficher le nombre d'observations dans chaque catégorie d'une variable catégorielle sous forme de barres. Il est particulièrement utile pour visualiser la distribution des données catégorielles. 

```python
plt.figure(figsize=(16,12))
ordre = df["League"].value_counts().index
sns.countplot(df, y="League", order=ordre)
plt.title("Nombre de joueurs par league")
plt.xlabel("Nombre de joueurs")
plt.show()
```

- La league "Sudamerica" a le plus grand nombre de joueurs, suivie de la league "MLS" et de la league "EFL Championship". Cela peut être dû à la popularité du football dans ces régions et à la présence de nombreux clubs professionnels qui recrutent des joueurs.

#### Quel pays (Nation) est le plus présent dans le dataset ?

```python
plt.figure(figsize=(12,8))
top = df["Nation"].value_counts().head(10)
sns.barplot(x=top.values, y=top.index, palette="viridis")
plt.xlabel("Nombre de joueurs")
plt.ylabel("Nation")
plt.title("Top 10 des nations avec le plus de joueurs")
plt.show()
```

## 3 · Bivarié

#### Y a-t-il une relation entre PAC (vitesse) et OVR ?

```python
plt.figure(figsize=(12,8))
sns.scatterplot(data=df, x="PAC", y="OVR", alpha=.4)
plt.title("Relation entre Vitesse (PAC) et Overall (OVR)")
print("Correlation PAC-OVR:\n",df[["PAC", "OVR"]].corr())
```

Le nuague du points monte que la vitesse et l'ovr sont globalement liés positivement, mais avec une dispersion assez large, donc y a beaucoup de joueurs s'écartent de cette tendance générale, surtout à faible Vitesse.

#### Faire une régression linéaire entre DRI et OVR.

```python
plt.figure(figsize=(12,8))
sns.regplot(data=df, x="DRI", y="OVR", scatter_kws={"alpha":0.4}, line_kws={"color":"red"})
plt.title("Relation entre Dribbling (DRI) et Overall (OVR) avec régression linéaire")
plt.xlabel("Dribbling")
plt.ylabel("Overall")
plt.show()
```

- La droite rouge monte qu'en moyenne, plus un joueur a un bon dribbling, plus sa note globale a tendance à être élevée, mais le lien reste modéré, pas parfait.

#### Utiliser un jointplot pour visualiser PAS vs DRI.

```python
plt.figure(figsize=(12,8))
sns.jointplot(data=df, x="PAS", y="DRI", kind="hex", palette="viridis") 
plt.title(" \n Distribution jointe entre Passes (PAS) et Dribbling (DRI)", y=0.95)
plt.show()
```

- on observeque la majorité des joueurs ont des niveaux de passes et de dribble moyens, avec une concentration plus importante autour de 60-70 pour les deux attributs. Il y a également quelques joueurs avec des niveaux de passes et de dribble exceptionnellement élevés, mais ils sont moins nombreux.

#### Faire un pairplot complet sur OVR, PAC, SHO, PAS, DRI.

- Le "pairplot" sert à visualiser les relations entre plusieurs variables quantitatives d'un DataFrame. Il crée une grille de graphiques où chaque variable est représentée sur l'axe des x et des y, permettant ainsi d'examiner les relations entre toutes les paires de variables. Cela peut aider à identifier des tendances, des corrélations ou des motifs dans les données.

```python
plt.figure(figsize=(12,8))
sns.pairplot(df[["OVR", "PAC", "SHO", "PAS", "DRI"]], corner=True, plot_kws={"alpha": .3})
plt.suptitle("Relations entre OVR, PAC, SHO, PAS et DRI", y=1.02)
plt.show()
```

#### Comparer les notes OVR par League avec un boxplot.

- on remarque que la distribution de OVR varie considérablement entre les ligues, avec certaines ligues ayant des joueurs avec des OVR plus élevés en moyenne que d'autres. Cela peut être dû à la qualité des joueurs dans chaque ligue, ainsi qu'à la compétitivité et à la popularité de chaque ligue.

#### Créer un violinplot de DRI par poste (Position), ou par ligue à défaut.

On utilise un violin plot pour visualiser la distribution des statistiques PAC, DRI et SHO pour les trois premières ligues du dataset. Le violin plot combine un boxplot et une estimation de densité, ce qui permet de visualiser à la fois la médiane, l'étendue et la forme de la distribution des données.

```python
plt.figure(figsize=(12,8))
ordre = df.groupby("Position")["DRI"].median().sort_values().index
sns.violinplot(df, x="Position", y="DRI", order=ordre, inner="quartile", cut=0)
plt.xlabel("DRI")
plt.ylabel("position")
plt.title("Distribution de DRI par position")  
plt.show()
```

- Les milieux de terrain (CDM, CM, CAM) ont tendance à avoir des valeurs de DRI plus élevées que les défenseurs (CB, LB, RB) et les attaquants (ST, CF). 
Cela est logique car les milieux de terrain sont souvent responsables de la création d'occasions et du contrôle du jeu, ce qui nécessite une meilleure capacité de dribble.

## 4 · Multivarié

#### Catplot comparant OVR par League, séparé par Nation (top 5 nations uniquement).

```python
plt.figure(figsize=(12,8))
top_5nations = df["Nation"].value_counts().index[:5]
sns.catplot(df[df["Nation"].isin(top_5nations)], x="OVR", y="League", col="Nation", kind="box", col_wrap=3, height=3.5)
plt.show()
```

#### lmplot pour visualiser PAC vs SHO, coloré par League.

```python
plt.figure(figsize=(12,8))
sns.lmplot(data=df[df["League"].isin(Top_league_5.index)],x="PAC",y="SHO",hue="League",scatter_kws={"alpha": 0.3})
plt.show()
```

- D'après le "linear model plot" (lmplot), on remarque une relation positive entre la vitesse et la précision de tir pour les joueurs des cinq ligues les plus représentées. Cela signifie que les joueurs plus rapides ont tendance à avoir une meilleure précision de tir. Cependant, il est important de noter que cette relation peut varier en fonction de la ligue et d'autres facteurs individuels des joueurs.

#### Heatmap de corrélation de toutes les stats numériques.

```python
num = df.select_dtypes(include="number").columns
corr = df[num].corr()
plt.figure(figsize=(20, 16))
sns.heatmap(corr,annot=True,fmt=".2f",cmap="RdBu_r",center=0,vmin=-1,vmax=1,mask=np.triu(np.ones_like(corr, dtype=bool), k=1))
plt.title("Matrice de corrélation des statistiques numériques")
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

- D'après la matrice de correlation, on remarque que les statistiques de vitesse (PAC) et de tir (SHO) sont corrélées positivement avec l'OVR, ce qui signifie que les joueurs ayant de meilleures statistiques dans ces domaines ont tendance à avoir un OVR plus élevé. En revanche, la corrélation entre l'OVR et d'autres statistiques telles que la défense (DEF) ou le physique (PHY) est moins prononcée, ce qui suggère que ces aspects du jeu ont moins d'impact sur l'évaluation globale des joueurs.
